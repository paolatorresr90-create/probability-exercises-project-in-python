## Ejercicio 1 

Dos dados se lanzan una vez y se observa el total obtenido. Usa una simulación para encontrar la probabilidad estimada de que la puntuación sea mayor a 7 o un número par. Una simulación es una repetición del mismo experimento multiples veces para observar su comportamiento:

- Ejecuta el experimento 1000 veces (lanza 2 dados 1000 veces, y suma el número de ambos dados).
- Lleva ua cuenta de los números y las veces que la suma fue mayor a 7 o un número par.
- Divide el número del paso 2 entre el número de iteraciones(1000).

```python
import random

# Configuración de la simulación
iteraciones = 1000
casos_exitosos = 0

for _ in range(iteraciones):
    # 1. Simular el lanzamiento de dos dados
    dado1 = random.randint(1, 6)
    dado2 = random.randint(1, 6)
    suma = dado1 + dado2
    
    # 2. Verificar las condiciones (Suma > 7 O Suma es par)
    # Usamos el operador % para saber si es par (residuo 0 al dividir por 2)
    if suma > 7 or suma % 2 == 0:
        casos_exitosos += 1

# 3. Calcular la probabilidad estimada
probabilidad_estimada = casos_exitosos / iteraciones

print(f"Resultados tras {iteraciones} lanzamientos:")
print(f"Casos que cumplieron la condición: {casos_exitosos}")
print(f"Probabilidad estimada: {probabilidad_estimada}")
```

## Ejercicio 2

Una caja contiene 10 bolas blancas, 20 bolas rojas y 30 bolas verdes. Si tomamos 5 bolas de la caja reemplazándolas (tomando una bola, anotando el color y luego regresandola a la caja). Queremos saber la probabilidad de:

1. Tomar 3 blancas y 2 rojas.
2. Toas todas del mismo color.

Ejecuta el experimento 1000 veces y calcula las probabilidades que mencionadas.

```python
import random

# Crear la caja con 10 blancas, 20 rojas y 30 verdes
ball_box = (['blanca'] * 10) + (['roja'] * 20) + (['verde'] * 30)

iteraciones = 1000
exitos_punto_1 = 0 # 3 blancas y 2 rojas
exitos_punto_2 = 0 # Todas del mismo color

for i in range(iteraciones):
    # random.choices permite tomar muestras "con reemplazo"
    muestra = random.choices(ball_box, k=5)
    
    blancas = muestra.count('blanca')
    rojas = muestra.count('roja')
    verdes = muestra.count('verde')
    
    # 1. Probabilidad de 3 blancas y 2 rojas
    if blancas == 3 and rojas == 2:
        exitos_punto_1 += 1
        
    # 2. Probabilidad de todas del mismo color
    if blancas == 5 or rojas == 5 or verdes == 5:
        exitos_punto_2 += 1

print(f"Probabilidad punto 1: {exitos_punto_1/iteraciones}")
print(f"Probabilidad punto 2: {exitos_punto_2/iteraciones}")
```
